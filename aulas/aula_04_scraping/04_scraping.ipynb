# Requests

A biblioteca `requests` é utilizada para fazer pedidos no protocolo **HTTP** e **HTTPS**, assim como outras ferramentes de autenticação, incluindo os pedidos comuns de `GET`, `POST` e outros.  
Como ela já vem instalada no sistema *Colab*, só é necessário importar:

````python
import requests
````

O uso mais básico de `requests` é pelo método `requests.get(url)`:

````python
url = 'https://www.eel.usp.br/ensino/graduacao/cursos' # Define o url a ser usado
response = requests.get(url) # Usa requests para pegar o conteúdo da página
response # Mostra a resposta
````

Um *output* de `<Response [200]>` corresponde a um OK, e que o pedido retornou o esperado. Caso ocorra outra resposta, como um 403, 404, comumente indica erro:



````python
url_inexistente = 'https://www.eel.usp.br/ensino/bombas_nucleares' # Este url não existe
response_erro = requests.get(url_inexistente) # Pega o conteúdo de uma página inexistente
response_erro # Mostra a resposta
````

````python
print(response.content) # Imprime todo o conteúdo
print('------------------------------------------')
print(response.text[:1000]) # Imprime os 1000 primeiros caracteres do .text
````

`.content` traz o conteúdo do tipo `bytes`, sem ser decodificado:

````python
type(response.content) # Mostra o tipo de response.content
````

````python
type(response.text) # Mostra o tipo de response.text
````

## APIs

Uma API (Interface de Programação de Aplicações) é um conjunto simplificado de regras com interfaces que permitem comunicar com o sistema, e para nós, sua capacidade mais importante é a de coleta de dados.  
APIs são especialmente úteis para *data mining*, que é o processo de aquisição e exploração de dados. Elas permitem acesso de grandes volumes, de forma automática.


## Exemplo 1

O IBGE possui diversas APIs, disponíveis gratuitamente junto com suas [documentações](https://servicodados.ibge.gov.br/api/docs/). Recomenda-se, assim como todas APIs, ler cautelosamente a documentação para aprender seu uso.

No *python*, comumente se retorna APIs pela biblioteca `requests`.

Vamos usar de exemplo a API de [publicações](https://servicodados.ibge.gov.br/api/docs/publicacoes):

````python
pesquisa = 'agro'
url_ibge = 'http://servicodados.ibge.gov.br/api/v1/publicacoes/' + pesquisa
````

Com o URL, é só utilizar de requests:

````python

resposta_ibge = requests.get(url_ibge)
resposta_ibge.content
````

Como a resposta veio em formato json, podemos converter:

````python
resposta_dict = resposta_ibge.json()
resposta_dict
````

Assim, ao conhecer sua estrutura, é possível converter em uma DataFrame:

````python
df_ibge = pd.DataFrame(resposta_dict)
df_ibge.head()
````

E até manipular em mais detalhes, como retornar apenas o nome de cada publicação:

````python
df_publicacoes = df_ibge['items'].apply(pd.Series)
df_publicacoes.head()
````

Segue outro exemplo básico de API, porém útil:  
A API do INMET - comunicada através do url (`https://apitempo.inmet.gov.br/token/estacao/`) - retorna `json` já apropriadamente formatado.

#### Explicação da API do INMET
##### Como adquirir um token
O cadastro é feito por e-mail para o endereço `cadastro.act@inmet.gov.br`, em que é necessário enviar as seguintes informações.

```
Curso de formação: curso...
Prazo necessário para a pesquisa: começo-fim.
Nome da instituição: Universidade de São Paulo, unidade de Lorena (EEL).
Tipo de instituição: PÚBLICA.
Finalidade de uso dos dados: Explicação da finalidade de uso.
Nome do responsável: Nome Sobrenome
E-mail do responsável: email@usp.br
```

Adapte para seu caso.

##### Como utilizar:
Você provavelmente recebeu uma mensagem deste tipo em seu e-mail, caso já adquiriu token:
- Os códigos das estações estão nos links:  
[Manuais](https://apitempo.inmet.gov.br/estacoes/M)  
[Automáticas](https://apitempo.inmet.gov.br/estacoes/T)


- Para recuperar dados horários referentes a estações automáticas ou manuais, segue o modelo:
`apitempo.inmet.gov.br/token/estacao/:data_inicial/:data_final/:cod_estacao/:seu_token`  
Por exemplo, para pegar dados de 2022-11-01 a 2022-12-03 da estação A001:
`https://apitempo.inmet.gov.br/token/estacao/2022-11-01/2022-11-01/A001/fbb192ba9f1a4673f6a62400943f391292d1d582fffbc5fd21ab5239a681929a`


- Para recuperar dados horários de TODAS as estações automáticas de um determinado dia e horário::
`apitempo.inmet.gov.br/token/estacao/dados/:data/:hora/:seu_token`
Por exemplo, pegar dados de 2022-11-01 às 00h00m: `https://apitempo.inmet.gov.br/token/estacao/dados/2022-11-01/0000/fbb192ba9f1a4673f6a62400943f391292d1d582fffbc5fd21ab5239a681929a`

- Para recuperar dados diários referentes a estação automática ou manual:
`apitempo.inmet.gov.br/token/estacao/diaria/:data_inicial/:data_final/:cod_estacao/:seu_token`
Por exemplo, pedir dados diários de 2022-11-01 às 2023-01-01 da estação A001: `https://apitempo.inmet.gov.br/token/estacao/diaria/2022-11-01/2023-01-01/A001/fbb192ba9f1a4673f6a62400943f391292d1d582fffbc5fd21ab5239a681929a`


Vamos tentar pegar os dados de 2022-11-01 da estação A001. Para isso, é conveniente colocar as diferentes partes do URL como diferentes variáveis:

````python
url_base = ' https://apitempo.inmet.gov.br/token/estacao'
data_inicial = '2022-11-01'
data_final = '2022-11-30'
estacao = 'A001'
api_key = 'INSIRA SUA CHAVE API AQUI'

# Junta as diferentes partes do url em uma única string:
url_final = url_base + '/' + data_inicial + '/' + data_final + '/' + estacao + '/' + api_key
print(url_final) # Imprime o URL final

resposta = requests.get(url_final) # Coleta os dados de url_final

resposta_dic = resposta.json() # Transforma o url em formato json, semelhante a um dict

print(resposta_dic) # Imprime o dict
````

Como a API já retorna um arquivo json, pode-se usar o `url_final` diretamente:

````python
df = pd.read_json(url_final)
df
````

Caso queremos usar `resposta_dic` para fazer a `DataFrame`, usamos o método: `pd.DataFrame.from_dict(dict_qualquer)`:

````python
df = pd.DataFrame.from_dict(resposta_dic)
df
````

### Exemplo 3

Vamos usar uma API um pouco mais complicada, que gera dados aleatórios em um formato que exige ser trabalhado:  
`url = https://randomuser.me/api`

````python
url = 'https://randomuser.me/api' # Salva o URL
resposta = requests.get(url) # Coleta o conteúdo do url
pprint.pprint(resposta.text) # Usa pprint para mostrar o texto da resposta
````

Como vemos, a resposta veio no formato de json, e podemos converter em um `dict`:

````python
resposta_dic = resposta.json() # Transforma a resposta em dict
print(type(resposta_dic)) # Mostra o type da variável resposta_dic
print(resposta_dic) # Mostra a resposta_dic
````

Ao analisar um pouco o dicionário, percebe-se que os dados que queremos estão sob a chave `results`:

````python
resposta_dic['results'] # Pega o valor da chave "results"
````

Porém, percebe-se que os dados ainda estão em uma lista, então, para ter o dicionário final:

````python
dic_final = resposta_dic['results'][0] # Pega o primeiro item do valor da chave "results"
dic_final # Mostra dic_final
````

E como ainda há dicionários dentro de dicionários, podemos usar a função `pd.json_normalize(variavel_json)`:

````python
var_json = resposta_dic['results'][0] # Retoma o dicionário (json)
dic_norm = pd.json_normalize(var_json) # Normaliza o dicionário
pd.DataFrame(dic_norm) # Transforma o dicionário normalizado em DataFrame
````

Também é interessante fazer todas estas etapas em apenas uma linha:

````python
pd.DataFrame(pd.json_normalize(resposta.json()['results'][0]))
````

# Web scraping

*Web scraping* é a extração automatizada de dados de sites. Para nós, seu maior interesse é na extração de dados tabelados, ou até não-tabelados, que não estão disponíveis em arquivos separados.   
Vamos usar de exemplo uma tabela de um html adquirida do sistema [TABNET](https://datasus.saude.gov.br/informacoes-de-saude-tabnet/) do Datasus.

````python
url_tabnet = "https://github.com/JCFDGG/LOB1266-PUB/raw/refs/heads/main/ExemplosAula/webscrap/tabnet.html"
caminho_tabnet = '../../dados/04_scraping/tabnet.html'

texto_tabnet = requests.get(url_tabnet).text # Pega o texto da resposta diretamente

pprint.pprint(texto_tabnet[:1000]) # Usa pprint para mostrar os dados
````

Vemos que não temos apenas a tabela, mas muitas informações desnecessárias para a análise. É possível isolar a tabela por métodos da classe `string`, mas é mais útil utilizar de `BeautifulSoup`:

## A biblioteca BeautifulSoup

A biblioteca `bs4` nos dá o método e classe `BeautifulSoup`. Então, importa-se assim:

````python
from bs4 import BeautifulSoup
````

Vamos salvar o processamento em uma variável de nome `soup`, que precisa da lógica:


```
soup = BeautifulSoup( texto_html_qualquer, 'html5lib')
```

`'html5lib'` é um parser, que vai lidar com o html em si, e permitir extrair informações dele. Há outros, como o [nativo](https://docs.python.org/3/library/html.parser.html) ou o [lxml](https://lxml.de/), mas o [html5lib](https://github.com/html5lib/) permite maior flexibilidade e melhor compatibilidade com html "quebrado".  
Segue um exemplo, utilizando o TABNET:

````python
soup = BeautifulSoup(texto_tabnet,'html5lib')
````

De uma forma similar ao `pprint`, O método `soup.prettify()` mostra formatado adequadamente:

### Método find e find_all

Uma das principais utilidades de BeautifulSoup é encontrar elementos HTML facilmente, através do método `.find(tag)` ou `.find_all(tag)`, em que a `tag` é qualquer uma das disponíveis por HTML, seguem como exemplos:

- Parágrafos, de tag 'p'


```
soup.find('p') # Encontra o primeiro parágrafo da variável soup
soup.find_all('p') # Encontra todos parágrafos da variável soup
```


````python
soup.find_all('p') # Encontra todos parágrafos, e retorna uma lista
````



- Links, de tag 'a'

```
soup.find('a') # Encontra o primeiro link da variável soup
soup.find_all('a') # Encontra todos links da variável soup
```



````python
soup.find('a') # Encontra o primeiro link
````

O método `soup.find_all()` retorna um set iterável, logo é possível pegar um link específico:

````python
soup.find_all('a')[2] # Pega o 3° link no documento HTML.
````

Note que também é possível extrair o texto:

````python
soup.find('p').text # Extrai o texto do primeiro parágrafo
````

#### Tabelas

Tabelas também podem ser descobertas pelo método .find(), mas elas normalmente precisam de um pouco de tratamento para serem utilizadas. A tag de tabelas é "table":

````python
tabela = soup.find('table') # Encontra a primeira tabela
````

O pandas possui um método para diretamente ler um HTML, `pd.read_html()`. Porém, ele só consegue ler do tipo `StringIO` (em versões mais antigas ele consegue ler `HTML` literal), o que nossa tabela não é:

````python
type(tabela) # Mostra o tipo que soup.find(...) traz
````

Mas, para modificar, podemos simplesmente encodifica-la:

````python
tabela_encoded = tabela.encode('utf-8') # Encodifica o HTML em utf-8
df = pd.read_html(tabela_encoded) # Lê todas tabelas do HTML
df[0] # O método retorna uma lista, pois espera que hajam várias tabelas. Pegamos o primeiro elemento da lista.
````

#### Exemplo de HTMLs com múltiplas tabelas

Podemos também especificar alguma restrição com um dicionário com outras chaves `HTML`, para quando há mais de uma tabela:

````python
# Abaixo está um HTML com duas tabelas, tente mudar a classe e ID das tabelas para ver o que ocorre.
html_exemplo = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Test Tables</title>
</head>
<body>

    <table class="Notas">
        <tr>
            <th>Nome</th>
            <th>Nota</th>
        </tr>
        <tr>
            <td>Alice</td>
            <td>9.9</td>
        </tr>
        <tr>
            <td>Heitor</td>
            <td>9.8</td>
        </tr>
        <tr>
            <td>Victor</td>
            <td>9.7</td>
        </tr>
    </table>

    <table id="id_qualquer">
        <tr>
            <th>Animal</th>
            <th>População</th>
        </tr>
        <tr>
            <td>Unicórnio</td>
            <td>27</td>
        </tr>
        <tr>
            <td>Dragão</td>
            <td>13</td>
        </tr>
        <tr>
            <td>Pé-grande</td>
            <td>1</td>
        </tr>
    </table>

</body>
</html>
"""
````

````python
soup_exemplo = BeautifulSoup(html_exemplo,'html5lib') # Salva o html processado na variável

tabela_classe = soup_exemplo.find('table', # Encontra a primeira tabela
 {'class':'Notas'}) # Cuja class = Notas


tabela_id = soup_exemplo.find('table', # Encontra a primeira tabela
 {'id':'id_qualquer'}) # Cujo id = id_qualquer

print(tabela_id.prettify()) # Imprime a tabela_id.
````

Também pode-se apenas ler todas a tabelas diretamente, e selecionar por *slicing*:

````python
dfs = pd.read_html(html_exemplo.encode('utf-8')) # Lê o HTML já encodificado, retorna uma lista

df_id = dfs[1] # Pega a segunda tabela da lista
df_class = dfs[0] # Pega a primeira tabela da lista
df_id
````
